```python
import os
import scanpy as sc
import pandas as pd
import anndata as ad
import numpy as np
```

```python
in_dir = '../../results/03_kidney_atlas/01_viz_data'
out_dir = '../../results/03_kidney_atlas/02_create_anndata'

os.makedirs(out_dir, exist_ok=True)
```

```python
adata = ad.io.read_mtx(f"{in_dir}/rna/matrix.mtx").T  # transpose the data
genes = pd.read_csv(f"{in_dir}/rna/genes.tsv", sep="\t", header=None)
adata.var_names = genes[0].values
adata.var["genes"] = genes[0].values
barcodes = pd.read_csv(f"{in_dir}/rna/barcodes.tsv", 
                       sep="\t", header=None)
# remove the perfix to match cell barcode in fragment file
adata.obs_names = barcodes[0].values
```

```python
# add UMAP
df_umap = pd.read_csv(f'{in_dir}/umap.csv', index_col=0)
df_umap = df_umap.loc[adata.obs_names]
adata.obsm["X_umap"] = df_umap.values
```

```python
df_obs = pd.read_csv(f'{in_dir}/meta.csv', index_col=0)
df_obs = df_obs.loc[adata.obs_names]
adata.obs = df_obs
```

```python
adata.obs
```

```python
sc.set_figure_params(figsize=(15, 15))
sc.pl.umap(adata, color='subclass.l2', legend_loc='on data')
```

```python
adata.obs['condition.l3'] = adata.obs['condition.l3'].astype(str)

sc.pl.umap(adata, color='condition.l3', legend_loc='on data')
```

```python
adata.obs['patient'] = adata.obs['patient'].astype(str)

sc.pl.umap(adata, color='patient', legend_loc='on data')
```

```python
adata.obs['location'] = adata.obs['location'].astype(str)

sc.pl.umap(adata, color='location', legend_loc='on data')
```

```python
adata.obs['laterality'] = adata.obs['laterality'].astype(str)

sc.pl.umap(adata, color='laterality', legend_loc='on data')
```

```python
adata.write_h5ad(f"{out_dir}/adata.h5ad")
```
