```r
suppressMessages(library(Seurat))
suppressMessages(library(SeuratDisk))
suppressMessages(library(dplyr))
suppressMessages(library(ggplot2))
suppressMessages(library(Matrix))
```

```r
out_dir <- "../../results/03_kidney_atlas/01_viz_data"

# create output directory if it doesn't exist
if (!dir.exists(out_dir)){
  dir.create(out_dir, recursive = TRUE)
}
```

```r
obj <- LoadH5Seurat("../../data/kidney_atlas_Nature2023/GSE183277_Kidney_Healthy-Injury_Cell_Atlas_snCv3_Seurat_03282022.h5Seurat")
```

```r
rna.counts <- obj@assays$RNA@counts
```

```r
dim(rna.counts)
```

```r
# save data
writeMM(rna.counts, paste(out_dir, "/rna/", "matrix.mtx", sep=""))
write(x = rownames(rna.counts), file = paste(out_dir, "/rna/", "genes.tsv",sep=""))
write(x = colnames(rna.counts),file = paste(out_dir, "/rna/", "barcodes.tsv",sep=""))
```

```r
write.csv(obj@meta.data, glue::glue("{out_dir}/meta.csv"))
```
