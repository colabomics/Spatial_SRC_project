```python
import os
import scanpy as sc
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
in_dir = '../../Data/score_gene_list'
out_dir = '../../results/08_gene_list/02_viz_gene_score'
data_dir = '../../results/08_gene_list/01_viz_gene_score'

os.makedirs(out_dir, exist_ok=True)
```

```python
sample_list = ['15837', '15852', '16512', '21022', '8262', 
               'B23-03794-5', 'B23-05785-7-1', 'B23-05785-7-2', 
               'X1842', 'X1845', 'X1899', 'X1912']

#sample_list = ['15837', '15852']
```

```python

for gene_set in ['IFN_genes_all', 
                 'ECM_COLLAGENS_genes',
                 'ECM_PROTEOGLYCANS_genes',
                 'HALLMARK_INTERFERON_ALPHA_RESPONSE',
                 'HALLMARK_INTERFERON_GAMMA_RESPONSE',
                 'Complement_score_genes',
                 'ECM_GLYCOPROTEINS_genes']:
    
    os.makedirs(f"{out_dir}/{gene_set}", exist_ok=True)
    
    _vmax = 0
    _vmin = 0
    for sample in sample_list:
        adata = sc.read_h5ad(f'{data_dir}/{sample}/adata.h5ad')

        if adata.obs[gene_set].quantile(0.95) > _vmax:
            _vmax = adata.obs[gene_set].quantile(0.95)

        # if adata.obs[gene_set].min() < _vmin:
        #     _vmin = adata.obs[gene_set].min()

    for sample in sample_list:
        adata = sc.read_h5ad(f'{data_dir}/{sample}/adata.h5ad')
        fig, ax = plt.subplots(figsize=(6, 6))
        sc.pl.spatial(adata, color=gene_set, 
                      vmax=_vmax, vmin=_vmin, show=False, ax=ax)
        fig.savefig(f"{out_dir}/{gene_set}/{sample}.png", dpi=300)
        plt.close(fig)
```
