```python
import os
import shutil
from spatialdata_io import visium
import tarfile

import warnings
warnings.filterwarnings('ignore')
```

```python
in_dir = "../../Data/10x_visium"
out_dir = "../../results/09_analysis_podocyte_niche/01_prepare_data"

os.makedirs(out_dir, exist_ok=True)
```

```python
sample_list = ['15837', '15852', '16512', '21022', '8262', 
               'B23-03794-5', 'B23-05785-7-1', 'B23-05785-7-2', 
               'X1842', 'X1845', 'X1899', 'X1912']

# sample_list = ['15837']
```

```python
for sample in sample_list:
    # copy fullres image files
    # if sample in ['B23-05785-7-1', 'B23-05785-7-2']:
    #     src_file = "../../Data/HE_and_fastq/B23-05785-7.tif"
    # else:
    #     src_file = f"../../Data/HE_and_fastq/{sample}.tif"
    # dest_file = f"{in_dir}/{sample}/{sample}.tif"

    # # if not os.path.exists(dest_file):
    # #     shutil.copy(src_file, dest_file)

    sdata = visium(path=f"{in_dir}/{sample}",
                   dataset_id=f"{sample}", 
                   fullres_image_file=f'{sample}.tif',
                   var_names_make_unique=True)
    
    if os.path.exists(f"{out_dir}/{sample}.zarr"):
        shutil.rmtree(f"{out_dir}/{sample}.zarr")

    sdata.write(f"{out_dir}/{sample}.zarr")

    with tarfile.open(f"{out_dir}/{sample}.zarr.tar.gz", "w:gz") as tar:
        tar.add(f"{out_dir}/{sample}.zarr", arcname=f"{sample}.zarr")

    # remove the uncompressed zarr directory
    shutil.rmtree(f"{out_dir}/{sample}.zarr")
```
