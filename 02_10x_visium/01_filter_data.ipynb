```python
import os
import warnings
warnings.filterwarnings("ignore")
import squidpy as sq
import scanpy as sc
import anndata as ad
```

```python
in_dir = "../../data/10x_visium"
out_dir = "../../results/02_10x_visium/01_filter_data"
os.makedirs(out_dir, exist_ok=True)
```

```python
sample_list = ['15837', '15852', '16512', '21022', '8262', 
               'B23-03794-5', 'B23-05785-7-1', 'B23-05785-7-2', 
               'X1842', 'X1845', 'X1899', 'X1912']
```

```python
for sample in sample_list:
    adata = sq.read.visium(f"{in_dir}/{sample}")

    adata.obs_names_make_unique()
    adata.obs['sample'] = f"sample_{sample}"

    sc.pp.calculate_qc_metrics(adata, inplace=True)
    sc.pp.filter_cells(adata, min_genes=500)

    sq.pl.spatial_scatter(adata, color='total_counts', title=f'Sample {sample}')

    # Save filtered data
    adata.write_h5ad(f"{out_dir}/{sample}.h5ad")
```
