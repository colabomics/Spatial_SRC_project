```python
import os
import pandas as pd
```

```python
df_de_ln = pd.read_csv('../../results/02_10x_visium/03_de_analysis/de_results_LN_vs_Normal.csv', 
                       index_col=0)
df_de_src = pd.read_csv('../../results/02_10x_visium/03_de_analysis/de_results_SRC_vs_Normal.csv', 
                       index_col=0)
```

```python
df_de_ln.head()
```

```python
df_de_src.head()
```

```python
in_dir = '../../manuscript/figures/main_fig1/upset_intersections'
out_dir = '../../results/02_10x_visium/08_sel_genes'

os.makedirs(out_dir, exist_ok=True)
```

```python
df = pd.read_csv(f"{in_dir}/LN_down.csv", header=None)
df = df_de_ln.loc[df_de_ln.index.isin(df[0]), :]
df.to_csv(f"{out_dir}/LN_down.csv")
```

```python
df = pd.read_csv(f"{in_dir}/LN_up.csv", header=None)
df = df_de_ln.loc[df_de_ln.index.isin(df[0]), :]
df.to_csv(f"{out_dir}/LN_up.csv")
```

```python
df = pd.read_csv(f"{in_dir}/LN_up_SRC_down.csv", header=None)
df1 = df_de_ln.loc[df_de_ln.index.isin(df[0]), :]
df2 = df_de_src.loc[df_de_src.index.isin(df[0]), :]
df1['comparison'] = 'LN_up'
df2['comparison'] = 'SRC_down'
df = pd.concat([df1, df2], axis=0)
df.to_csv(f"{out_dir}/LN_up_SRC_down.csv")
```

```python
df = pd.read_csv(f"{in_dir}/SRC_down.csv", header=None)
df = df_de_src.loc[df_de_src.index.isin(df[0]), :]
df.to_csv(f"{out_dir}/SRC_down.csv")
```

```python
df = pd.read_csv(f"{in_dir}/SRC_up.csv", header=None)
df = df_de_src.loc[df_de_src.index.isin(df[0]), :]
df.to_csv(f"{out_dir}/SRC_up.csv")
```

```python
df = pd.read_csv(f"{in_dir}/SRC_down_LN_down.csv", header=None)
df1 = df_de_ln.loc[df_de_ln.index.isin(df[0]), :]
df2 = df_de_src.loc[df_de_src.index.isin(df[0]), :]
df1['comparison'] = 'LN_down'
df2['comparison'] = 'SRC_down'
df = pd.concat([df1, df2], axis=0)
df.to_csv(f"{out_dir}/SRC_down_LN_down.csv")
```

```python
df = pd.read_csv(f"{in_dir}/SRC_up_LN_up.csv", header=None)
df1 = df_de_ln.loc[df_de_ln.index.isin(df[0]), :]
df2 = df_de_src.loc[df_de_src.index.isin(df[0]), :]
df1['comparison'] = 'LN_up'
df2['comparison'] = 'SRC_up'
df = pd.concat([df1, df2], axis=0)
df.to_csv(f"{out_dir}/SRC_up_LN_up.csv")
```
