```python
import os
import pandas as pd
```

```python
in_dir = '../../results/07_scTM/01_run_scTM'
out_dir = '../../results/07_scTM/02_get_topic_markers'

os.makedirs(out_dir, exist_ok=True)
```

```python
df = pd.read_csv(f"{in_dir}/scTM_15topics_beta.csv", index_col=0)
```

```python
df.head()
```

```python
# select top 200 genes for each column in df
markers = {}
for col in df.columns:
    df_sort = df.sort_values(by=col, ascending=False)

    markers[col] = df_sort.head(n = 200).index
```

```python
df_markers = pd.DataFrame(markers)
```

```python
df_markers.to_csv(f"{out_dir}/topic_markers.csv", index=False)
```
